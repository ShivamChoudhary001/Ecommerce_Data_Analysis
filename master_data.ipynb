```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
# Loading the data
df_orders = pd.read_csv("raw_data/olist_orders_dataset.csv")
df_items = pd.read_csv("raw_data/olist_order_items_dataset.csv")
```

```python
# Merging the data
master_df = pd.merge(df_orders, df_items, on='order_id', how='inner')
```

```python
# To check the shape of data (Total Rows, Total Columns)
master_df.shape
```

```python
# To describe about the data
master_df.describe()
```

```python
# To check information about data columns and data-type of columns
master_df.info()
```

```python
# To change or correct the data type of columns like(date to datetime)
cols = ['order_purchase_timestamp', 'order_delivered_carrier_date', 'order_delivered_customer_date', 'order_estimated_delivery_date', 'shipping_limit_date']

for c in cols:
    master_df[c] = pd.to_datetime(master_df[c])
```

```python
# check again the info.
master_df.info()
```

```python
# To check the null values (missing values)
master_df.isnull().sum()
```

```python
# To filter the rows of missing dates.
missing_deliveries = master_df[master_df['order_delivered_customer_date'].isnull()]

# To check the status of missing orders.
print(missing_deliveries['order_status'].value_counts())
```

```python
# To connect with sql database
from sqlalchemy import create_engine

# Creating engine
engine = create_engine("sqlite:///ecommerce_data_analysis.db")

# To save master table into SQL
master_df.to_sql('powerbi_master_sales', con=engine, if_exists='replace', index=False)

print("Data Successfully Saved")
```

# Now we will anlayze data by SQL queries

```python
# To display table nice.
from IPython.display import display

# Query 1 :- To show 'Monthly Sales Trend
query_1 = """
Select strftime('%Y', order_purchase_timestamp) AS Year, 
CASE strftime('%m', order_purchase_timestamp)
    WHEN '01' THEN 'Jan'
    WHEN '02' THEN 'Feb'
    WHEN '03' THEN 'Mar'
    WHEN '04' THEN 'Apr'
    WHEN '05' THEN 'May'
    WHEN '06' THEN 'Jun'
    WHEN '07' THEN 'Jul'
    WHEN '08' THEN 'Aug'
    WHEN '09' THEN 'Sep'
    WHEN '10' THEN 'Oct'
    WHEN '11' THEN 'Nov'
    WHEN '12' THEN 'Dec'
END AS Month,
SUM(Price) as Total_Revenue
FROM powerbi_master_sales
GROUP BY Year, Month
ORDER BY Total_Revenue DESC;
"""

display(pd.read_sql(query_1, con=engine))
```

```python
# To check the highest quantity order and value.
query_2 = """
SELECT order_id, COUNT(product_id) AS Total_Qty, SUM(Price) AS Total_Value
FROM powerbi_master_sales
GROUP BY order_id
ORDER BY Total_Qty DESC
LIMIT 5;
"""

display(pd.read_sql(query_2, con=engine))
```

```python
# To check the average delivery time in days per delivered order.
query_3 = """
SELECT order_status, ROUND(AVG(julianday(order_delivered_customer_date) - julianday(order_purchase_timestamp)),2) AS Avg_delivery_time_in_days
FROM powerbi_master_sales
WHERE order_status = 'delivered'
GROUP BY order_status
;
"""

display(pd.read_sql(query_3, con=engine))
```

```python
# To check total_shipping_value ('freight_value) against total_sales_value
query_4 = """
SELECT order_status, SUM(Price) AS Total_Sales, SUM(freight_value) AS Total_Delivery_Charge
FROM powerbi_master_sales
GROUP BY order_status
ORDER BY Total_Delivery_Charge DESC;"""

display(pd.read_sql(query_4, con=engine,))
```
